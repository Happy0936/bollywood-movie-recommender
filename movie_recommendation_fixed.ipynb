```python
movies = movies_full[
    [
        'title_x',
        'genres',
        'story',
        'summary',
        'actors'
    ]
].copy()

movies.head()
```

```python
print("Before removing missing values:", movies.shape)

movies.dropna(inplace=True)

print("After removing missing values:", movies.shape)
```

```python
def convert_genres(text):
    return text.split('|')

movies['genres'] = movies['genres'].apply(convert_genres)
```

```python
movies[['title_x', 'genres']].head()
```

```python
def convert_actors(text):
    return text.split('|')

movies['actors'] = movies['actors'].apply(convert_actors)
```

```python
movies[['title_x','genres', 'actors']].head()
```

```python
movies['story'] = movies['story'].apply(lambda x: x.split())
```

```python
movies['summary'] = movies['summary'].apply(lambda x: x.split())
```

```python
movies['genres'] = movies['genres'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)

movies['actors'] = movies['actors'].apply(
    lambda x: [i.replace(" ", "") for i in x]
)
```

```python
movies['tags'] = (
    movies['story']
    + movies['summary']
    + movies['genres']
    + movies['actors']
)
```

```python
movies[['title_x', 'tags']].head()
```

```python
#NLP Process  Date: 4/09/26
```

```python
new_df = movies[['title_x', 'tags']].copy()

new_df.rename(columns={'title_x': 'title'}, inplace=True)

new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(
    lambda x: " ".join(x)
)
```

```python
new_df.head()
```

```python
new_df['tags'] = new_df['tags'].apply(
    lambda x: x.lower()
)
```

```python
new_df.head()
```

# Stemming


```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

def stem(text):

    y = []

    for word in text.split():
        y.append(ps.stem(word))

    return " ".join(y)


new_df['tags'] = new_df['tags'].apply(stem)

new_df.head()
```

```python
metadata = movies_full.loc[
    movies.index,
    [
        'genres',
        'imdb_rating',
        'imdb_votes',
        'year_of_release',
        'runtime',
        'poster_path'
    ]
].reset_index(drop=True)
```

```python
new_df.reset_index(drop=True, inplace=True)
```

```python
final_df = new_df.copy()

final_df = final_df.join(metadata)

final_df.head()
```

```python
print(final_df.columns.tolist())
```

```python
print("Final Dataset Shape:", final_df.shape)
```

```python
# Vectorization
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
    max_features=5000,
    stop_words='english'
)

vectors = cv.fit_transform(
    final_df['tags']
).toarray()

print("Vector Shape:", vectors.shape)
```

```python
feature_names = cv.get_feature_names_out()

word_counts = vectors.sum(axis=0)

top_indices = word_counts.argsort()[-20:][::-1]

top_words = feature_names[top_indices]
top_counts = word_counts[top_indices]

plt.figure(figsize=(10, 8))

plt.bar(top_words, top_counts)

plt.title("Top 15 Most Frequent Words in Movie Dataset")
plt.xlabel("Words")
plt.ylabel("Frequency")

plt.xticks(rotation=50)

plt.show()
```

```python
# Cosine Similarity
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)

print("Similarity Matrix Shape:", similarity.shape)


#C=A.B/(|A|.|B|)
```

```python
def recommend(movie):
    matches = final_df.index[final_df['title'] == movie].tolist()
    if not matches:
        raise ValueError(f'Movie not found: {movie}')
    movie_index = matches[0]
    distances = similarity[movie_index]
    ranked = sorted(
        ((i, score) for i, score in enumerate(distances) if i != movie_index),
        key=lambda item: item[1], reverse=True
    )
    return ranked[:5]
```

```python
movie = 'Uri: The Surgical Strike'
movie_info = final_df[final_df['title'] == movie]
if movie_info.empty:
    print(f'Movie not found: {movie}')
else:
    selected = movie_info.iloc[0]
    for field in ['title', 'genres', 'imdb_rating', 'imdb_votes', 'year_of_release', 'runtime', 'poster_path']:
        print(f'{field}: {selected[field]}')
```

```python
recommendations = recommend(movie) if movie in final_df['title'].values else []
```

```python
for i in recommendations:

    movie_data = final_df.iloc[i[0]]

    print("Title:", movie_data['title'])
    print("Genre:", movie_data['genres'])
    print("Rating:", movie_data['imdb_rating'])
    print("Votes:", movie_data['imdb_votes'])
    print("Year:", movie_data['year_of_release'])
    print("Runtime:", movie_data['runtime'])
    print("Poster:", movie_data['poster_path'])
    print("-" * 50)
```

```python
# Save the final movie data and similarity matrix for Streamlit
import pickle
from pathlib import Path

output_dir = Path.cwd()
with open(output_dir / "movie_list.pkl", "wb") as f:
    pickle.dump(final_df, f)
with open(output_dir / "similarity.pkl", "wb") as f:
    pickle.dump(similarity, f)
print("Saved both files in:", output_dir)
```
